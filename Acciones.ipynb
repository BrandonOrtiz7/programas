```python
import pandas as pd
import numpy as np

# 1. En Google Colab, al subir los archivos directo, la ruta es la raíz limpia
# No necesitas poner discos locales (C:) porque ya estamos en la nube
tickers = ["NVDA", "WMT", "GLD", "JPM", "GEHC"]

precios_dict = {}

# 2. El bucle abre los archivos que arrastraste al panel izquierdo
for t in tickers:
    try:
        # Busca el archivo directo por su nombre en el panel de Colab
        ruta_archivo = f"{t}.csv"

        # Leemos usando 'Fecha' como renglones
        df_local = pd.read_csv(ruta_archivo, index_col='Fecha', parse_dates=True)

        # Guardamos la columna 'Cierre'
        precios_dict[t] = df_local['Cierre']
        print(f"✅ Archivo {t}.csv cargado correctamente.")

    except FileNotFoundError:
        print(f"❌ Error: No encontré el archivo '{t}.csv' en el panel izquierdo. Revisa el nombre.")

# 3. Juntamos todo y limpiamos datos faltantes
datos_precios = pd.DataFrame(precios_dict).sort_index().dropna()

# 4. PASO 3: Transformamos los precios a rendimientos porcentuales diarios
rendimientos_escenarios = datos_precios.pct_change().dropna()

# ¡Listo! Mostramos el resultado en pantalla
print("\n🔥 ¡MATRIZ DE ESCENARIOS CONECTADA EN LA NUBE! 🔥")
print(rendimientos_escenarios.head())
print(f"\nTienes un total de {len(rendimientos_escenarios)} escenarios listos para el optimizador lineal.")
```
